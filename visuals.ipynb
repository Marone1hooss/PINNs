```python
import torch
import matplotlib.pyplot as plt
import numpy as np

# Load the model
def load_and_visualize_model():
    # Load the saved model
    model_path = "/home/marwan.housni/lustre/manapy-um6p-st-msda-1wabcjwe938/users/marwan.housni/internship/PINNs/PINNsformer/TrainedModels/PINNsformer_model_L500_20250615_200939_201_pts.pth"
    model_state = torch.load(model_path)
    
    # Print model structure
    print("Model structure:")
    for key in model_state.keys():
        print(f"\nLayer: {key}")
        print(f"Shape: {model_state[key].shape}")
    
    # Visualize weights distribution
    plt.figure(figsize=(15, 10))
    for i, (name, weights) in enumerate(model_state.items()):
        if 'weight' in name:  # Only visualize weight matrices
            plt.subplot(3, 2, i+1)
            weights_flat = weights.flatten().detach().numpy()
            plt.hist(weights_flat, bins=50)
            plt.title(f'Weight Distribution: {name}')
            plt.xlabel('Weight Value')
            plt.ylabel('Frequency')
    
    plt.tight_layout()
    plt.show()

if __name__ == "__main__":
    try:
        load_and_visualize_model()
    except Exception as e:
        print(f"Error loading model: {e}")
```
